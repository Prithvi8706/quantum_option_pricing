# 03 — Quantum Option Pricing via Amplitude Estimation

## Theory

### Quantum Amplitude Estimation (QAE)

Quantum Amplitude Estimation (Brassard et al., 2002) is a generalisation of Grover search that estimates the probability $a = \sin^2(\theta)$ of a marked outcome in a quantum circuit. For option pricing, the algorithm proceeds in three steps:

#### 1. State Preparation — encoding the risk-neutral distribution

A unitary $\mathcal{A}$ loads a discretised log-normal distribution over $n$ qubits:

$$\mathcal{A}|0\rangle^n = \sum_{i=0}^{2^n - 1} \sqrt{p_i}\,|i\rangle$$

where $p_i$ is the risk-neutral probability that $S_T$ lands in bin $i$.

#### 2. Payoff Encoding

An ancilla qubit is rotated by an angle proportional to the option payoff $f(S_i) = \max(S_i - K, 0)$:

$$|i\rangle|0\rangle \xrightarrow{\mathcal{R}} |i\rangle\!\left(\cos(c\,f_i)|0\rangle + \sin(c\,f_i)|1\rangle\right)$$

The amplitude of the $|1\rangle$ ancilla state is then $a \approx c \cdot \mathbb{E}[f(S_T)]$, proportional to the option price.

#### 3. Amplitude Estimation

QAE uses $M$ queries to the Grover operator $\mathcal{Q} = \mathcal{A}(I - 2|0\rangle\langle 0|)\mathcal{A}^\dagger(I - 2|\Psi_1\rangle\langle\Psi_1|)$ and Quantum Phase Estimation to estimate $\theta$ (and hence $a$) with error:

$$|\hat{a} - a| \leq \frac{2\pi\sqrt{a(1-a)}}{M} + \frac{\pi^2}{M^2} = O\!\left(\frac{1}{M}\right)$$

This is a **quadratic speedup** over classical Monte Carlo, which achieves $O(1/\sqrt{N})$ with $N$ samples.

### Qiskit Finance Implementation

`qiskit_finance.circuit.library.EuropeanCallExpectedValue` constructs $\mathcal{A}\cdot\mathcal{R}$ automatically given a `LogNormalDistribution` and strike $K$. `qiskit_algorithms.AmplitudeEstimation` (or the faster `IterativeAmplitudeEstimation`) wraps the full QPE routine.
